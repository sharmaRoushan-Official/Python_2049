### 4. Tuples in Python

1. What is a Tuples?

A `tuple`a collection of data types in python used to store multiple values in a single variable.

### key features:

* Ordered Collections
* Immutable(Cannot be changed after creation)
* Allow duplicate
* Can store different data types
* Faster than lists
* Written using round brackets ()

```python
num = []  # mutable
print(num,type(num))
```

```python
tup1 = ()  # immutable
print(tup1,type(tup1))
```

### 2. Creating tuples

###### Method:1 - Using Parentheses


```python
fruits = ("apple",'banana',"mango")
print(fruits)
```

#### Method 2: Without parentheses

```python
colors = "red","green","blue"
print(colors)
```

#### Method 3: Using Tuple Constructor

tuple()

```python
numbers = tuple([10,25])
print(numbers)
```

```python
numbers = tuple((10,25))
print(numbers)
```

```python
numbers = tuple({10,25})
print(numbers)
```

### 3. Tuple with Different Data Types

Tuples can store multiple data types


```python
data = ("John",10,25.63,True,False)
print(data,type(data))
```

```python
print(data[0],type(data[0]))
print(data[1],type(data[1]))
print(data[2],type(data[2]))
print(data[3],type(data[3]))
print(data[4],type(data[4]))
```

### 6. Slicing of Elements
Used to extract a protion of the tuple:

syntax:

----
var[start:stop]

```python
fruits = ("Strawberry","apple","mango")
fruits
```

```python
fruits[2]  # indexing
```

```python
## slicing of elements

fruits[2:]
```

```python
fruits[1:200]
```

### Striding of tuple

----
Syntax:

tuple[start:end:skip]

```python
fruits = ("apple","mango","banana","strawberry","grapes","orange")
print(fruits)
```

```python
fruits[::2]
```

```python
fruits[::3]
```

#### 7. Tuple Immutability
Tuple cannot be modified after creating


```python
numbers = (10,20,30) # tuple is immutable
print(numbers)
```

```python
numbers[0] = "Vrushali"
print(numbers)
```

```python
numbers = [10,20,30] # list is mutable
print(numbers)
```

```python
numbers[0] = "Vrushali" # string is immutable
print(numbers)
```

```python
x = "Vrushali"
print(x)
```

```python
x[0] = "P"
x
```

### 8. Changing Tuple Elements:
Since tuple are immutable, convert into list:

```python
numbers = (10,20,30)
print(numbers)

changeValue = list(numbers)
print(changeValue)
changeValue[0] = "Vrushali"
print(changeValue)


number1 = tuple(changeValue)
print(number1)
```

### 10. Tuple Packing
Packing means storing multiple values into a tuple


```python
data = ("John",25,"Engineer") # packing
print(data)
```

### 11. Tuple Unpacking



```python
name,age,job = ("John",25,"Engineer") # packing
print(name)
print(age)
print(job)
```

### 12. Extended Unpacking

Using * operator (Asterisk)


```python
name,*job = ("John",25,"Engineer") # packing
print(name)
print(job)
```

```python
*name,job = ("John",25,"Engineer") # packing
print(name)
print(job)
```
