### 5. Sets in Python
-----

##### 1. Introduction to Sets

A `Set` in Python is a collection data type used to store multiple unique elements. 

It is based on matimatical concept of set. 

#### Key Charactersistics

1. Unordered Collection
2. Unidexed
3. Mutable (can add/remove items)
4. Does not allow duplicates value
5. Elements must be immutable(hashable)


```python
s = {} # by default dictionary
print(s,type(s))
```

```python
s1 = {1} 
print(s1,type(s1))
```

```python
### empty set 

s2 = set()
print(s2,type(s2))
```

### 2. Creating Sets
---
Method 1: Using Curly braces {}

```python
s1 = {1,2,3,4,5,5,5,60,70,-10}
print(s1)
```

### Method 2: Using Set Constructor


```python
s = set([10,10,10,12,14,15,2,2,2,3,3,7,8,9,-10,-9,-8])
s
```

### 3. Duplicate Elements
Sets automatically remove duplicates value.

```python
s1 = {10,20,20,30,40,50,60}
print(s1)
```

### 4. Data Types allowed in Sets

Elements must be immutable(hashable)

1. int
2. float
3. string
4. tuple
5. boolean
6. frozenset()


```python
s1 = {10,55.63,"hello",(10,20),True,frozenset((10,20))}
print(s1)
```

### Not Allowed:

Mutable Data types
* List
* dictionary
* set

```python
s1 = {10,55.63,"hello",(10,20),True,["hello"]}
s1
```

```python
s1 = {10,55.63,"hello",(10,20),True,{"hello"}}
s1
```

```python
s1 = {10,55.63,"hello",(10,20),True,{'a':'apple'}}
s1
```
