### 5. Sets in Python
-----

##### 1. Introduction to Sets

A `Set` in Python is a collection data type used to store multiple unique elements. 

It is based on matimatical concept of set. 

#### Key Charactersistics

1. Unordered Collection
2. Unidexed
3. Mutable (can add/remove items)
4. Does not allow duplicates value
5. Elements must be immutable(hashable)


````python
s = {} # by default dictionary
print(s,type(s))
````

````python
s1 = {1} 
print(s1,type(s1))
````

````python
### empty set 

s2 = set()
print(s2,type(s2))
````

### 2. Creating Sets
---
Method 1: Using Curly braces {}

````python
s1 = {1,2,3,4,5,5,5,60,70,-10}
print(s1)
````

### Method 2: Using Set Constructor


````python
s = set([10,10,10,12,14,15,2,2,2,3,3,7,8,9,-10,-9,-8])
s
````

### 3. Duplicate Elements
Sets automatically remove duplicates value.

````python
s1 = {10,20,20,30,40,50,60}
print(s1)
````

### 4. Data Types allowed in Sets

Elements must be immutable(hashable)

1. int
2. float
3. string
4. tuple
5. boolean
6. frozenset()


````python
s1 = {10,55.63,"hello",(10,20),True,frozenset((10,20))}
print(s1)
````

### Not Allowed:

Mutable Data types
* List
* dictionary
* set

````python
s1 = {10,55.63,"hello",(10,20),True,["hello"]}
s1
````

````python
s1 = {10,55.63,"hello",(10,20),True,{"hello"}}
s1
````

````python
s1 = {10,55.63,"hello",(10,20),True,{'a':'apple'}}
s1
````

### 10. Set Operations(Most Important Points)

Python sets support mathematical operations

1. Union
2. Intersection
3. Different
4. symmetric_different

````python
A = {1,2,3}
B = {3,4,5}

print(A)
print(B)
````

#### Union: Unique Value

1. Union

```
Method - .union()
symbol - |
```

````python
print(A)
print(B)
````

````python
# Method 
print(A.union(B))
````

#### Symbol



````python
print(A|B)
````

### 2. Intersection

Common elements 

```
Method: .intersection()
symbol: & [Ampersand]

```


````python
A = {1,2,3}
B = {3,4,5}

print(A)
print(B)
````

````python
### method
print(A.intersection(B))
````

````python
# symbol
print(A & B)
````

#### 3. Difference
```
Method - .difference()
symbol - -
```

A-B
B-A

````python
A = {1,2,3}
B = {3,4,5}

print(A)
print(B)
````

````python
# method 

# A-B

print(A.difference(B))
print(B.difference(A))
````

````python
# symbol 

print(A-B)
print(B-A)
````

### 4. symmetric_difference

Elements in either set but not both

Method = .symmetic_difference()

symbol = ^ [Caret]

````python
A = {1,2,3}
B = {3,4,5}

print(A)
print(B)
````

````python
# method 

A.symmetric_difference(B)
````

````python
# symbol 

A^B
````

### 11. Set Comparison Operations


````python
A = {1,2}
B = {1,2,3}
print(A)
print(B)
````

````python
print(A.issubset(B))
````

````python
print(A.issuperset(B))
````

````python
print(B.issuperset(A))
````

````python
print(B.issubset(A))
````

### Disjoint
Check if not common elements

````python
A = {1,2}
B = {3,4}
print(A.isdisjoint(B))
````

````python
A = {1,2,3}
B = {3,4}
print(A.isdisjoint(B))
````

### 12. Frozenset

A frozenset is an immutable version of set Once created, it cannot be modified.

````python
fs = frozenset({10,23,32,23})
print(fs)
````

````python
fs = frozenset([10,23,32,23])
print(fs)
````

````python
fs = frozenset((10,23,32,23))
print(fs)
````

````python
fs.add(30)
````
