A `Conditional Statement` allow a program to make decision based conditions(True/False).

#### 2. Basic Syntax

```

    code execute if condition is True

else:
    Code execute if conditon is False
```

#### Nested Conditions

condition inside conditions

````python
marks = int(input("Enter your marks: "))

# A - 90-100 , B- 80-90, C- 70-80, D - 60-70 , E- <60

if marks>0 and marks<=100:
    if marks>=90 and marks<100:
        print("Grade A")
    elif marks>=80 and marks<90:
        print("Grade B")
    elif marks>=70 and marks<80:
        print("Grade C")
    elif marks>60 and marks<70:
        print("Grade D")
    elif marks<60:
        print("Grade E")
    else:
        print("Invalid Marks")
else:
    print("Marks is in positive value belongs 0 to 100")
````

````python
#8. Write a Python program to determine whether a given year is a *leap year* or not.
num = float (input("Enter a year: "))
if num %4 == 0:
    print("The given year is leap year")
else:
    print("The given year is not a leap year")
````

```
To calculate whether a year is a leap year in the Gregorian calendar, follow these three simple rules:
1. Divisible by 4: If the year can be divided evenly by 4, go to step 2. If not, it is not a leap year (365 days).
2. Divisible by 100: If the year can also be divided evenly by 100, go to step 3. If it is not divisible by 100, it is a leap year (366 days, with February 29).
3. Divisible by 400: If the year can be divided evenly by 400, it is a leap year. Otherwise, it is not a leap year.

Quick Examples

• 2024: Divisible by 4, not by 100 → Leap year (366 days)
• 1900: Divisible by 4 and by 100, but not by 400 → Not a leap year (365 days)
• 2000: Divisible by 4, by 100, and by 400 → Leap year (366 days)
• 2025: Not divisible by 4 → Not a leap year (365 days)

```
