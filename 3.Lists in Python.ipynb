#### 7. Removing an elements from the list
```
A). remove(): Removes by value
B). pop(n): n is the index value, Remove by index value.
C). pop(): Removes by default last value.
D). del(n): n is index value, remove indexed value.
E). del(var): list var then it will deleted complete list 
F). clear(): Removes all elements and return empty list

````python
nums = [10,20,30,40,50]
print(nums)
````

````python
nums.remove(10) # by value
nums
````

````python
nums.pop(2) # by index value
nums
````

````python
nums.pop()  # by default last value
nums
````

````python
del(nums[1])
nums
````

````python
del(nums)
nums
````

````python
nums = [10,20,30,40,50]
print(nums)
````

````python
nums.clear()
nums
````

### 8. List methods

-----

|Method|work|
|---|----|
|`append`|Add item at the end|
|`extend`|Add mutiple items|
|`insert`|insert the index position|
|`remove(x)`| Remove x value|
|`pop(n)`| pop element n and n is index value|
|`pop()`|pop last element|
|`clear()`| removes all elements|
|`del(n)`| delete items of n position|
|`index(x)`| return index of value x|
|`count(x)`| counts occurrence of x value|
|`sort()`|sort list:asc|
|`reverse()`|Reverse the list|
|`copy()`| Return the shallow copy|

### 9. Index(x) and count(x)


````python
nums = [10,20,30,10,40,50,30,30]

print(nums)
````

````python
print(nums.index(10))
print(nums.index(20))
````

````python
print(nums.count(10))
print(nums.count(20))
print(nums.count(30))
````
