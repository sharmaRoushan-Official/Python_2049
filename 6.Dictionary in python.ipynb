### 6. Dictionary in Python

1. What is a Dictionary in Python?

A `dictionary` is a collection of key:value paris.

* Each key is unique.
* Each key maps to a value.
* Dicationary are mutable but it's key is immutable.
* it is Written in curly braces.


````python
d1 = {}
print(d1,type(d1))
````

````python
s1 = set()
print(s1)
print(type(s1))
````

````python
d1 = {1}
print(d1,type(d1))
````

````python
d1 = {"name":"Yogesh"}
print(d1,type(d1))
````

````python
student = {"name":"Arnav","age":17,"course":"Python"}
print(student)
````

````python
student = {
    "name":"Arnav",
    "age":17,
    "course":"Python"
}
print(student)
````

### 3. Charactersitics of Dictionary

Important Points

### 1. Key must be unique[Immutable]

````python
data = {
    "name":"Arnav",
    "name":"yogesh",
    "name":"Mayuri"
}
print(data)
````

### 2. Keys must be immutable

Allowed:

* string
* number
* tuple
* boolean: True, False


Not Allowed:
* List
* Set
* dictionary


````python
data = {
    11:"Python",
    "name":"Aarya",
    ("Hello world",100):"Tuple Key",
    True : "Boolean",
    False: "Boolean"
}
print(data)
````

### 3. Values be anything

Values can be:

* String
* Number
* list
* dictionary
* set
* tuple
* object

````python
data = {
    "name":"Arnav",
    "marks":[80,90,100,80],
    "address":{
        "city":"Pune",
        "state":"Maharashtra"
    }
}
print(data)
````

### 12. Nested Dictionary

###### Dictinary inside another dicdtionary

````python
students = {
    "student1":{
        "name":"Arnav",
        "age":17
    },
    "student2":{
        "name":"Vrushali",
        "age":45
    },
    "student3":{
        "name":"yogesh",
        'age':10
    }
}
print(students)
````

````python
students
````

````python
students['student3']['age'] = 40
students
````

````python
students['student2']['age'] = 14
students
````

````python
students['student1']["name"] = "Yash"
````

````python
students
````
