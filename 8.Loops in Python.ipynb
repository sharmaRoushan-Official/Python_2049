### What is For Loop?

A for loop is used to iterate over a sesquence(such as list,tuple, string, dictionary, number, set, range) and execute block of code repeatedly.


### Flow
1. Python takes first item from the sequence.
2. Assign a loop variable
3. Exeucte the  the block of code
4. Move the next item
5. Repeat unitll all items are processed.


```
for var in sequence:
    block of code
```

````python
fruits = ["Mango","Dragon Fruit","Strawberry","papaya","watermelon","banana","kiwi"]

for fruit in fruits:
    print(fruit)
````

````python
### Q4. Sum of Natural Numbers
# Write a Python program to calculate the sum of the first N natural numbers.

num = int(input("Enter any number: "))

sumValue = 0
for n in range(1,num+1):
    sumValue +=n
    # print(n)
print(f"Sum of the First {num} natural sum Value is:",sumValue)
````

````python
### 11. Count Consonent

# Write a Python program to count the total number of consonants in a given string 

name = "Yogesh Varpe"

vowel = "AEIOUaeiou"
totalVowel = 0
consoenentCount =0
for ch in name:
    if ch == " ":
        continue
    if ch in vowel:
        totalVowel+=1
    # print(ch)
    else:
        consoenentCount+=1
print("Total Vowel Chacters is:",totalVowel)
print("Total Consonent Chracters is:",consoenentCount)
````

````python
name = "Yogesh Varpe"

vowel = "AEIOUaeiou"
totalVowel = 0
consoenentCount =0
i=0
while i<len(name):
    if name[i] == " ":
        i+=1
        continue
    if name[i] in vowel:
        totalVowel+=1
    # print(ch)
    else:
        consoenentCount+=1
    i+=1
print("Total Vowel Chacters is:",totalVowel)
print("Total Consonent Chracters is:",consoenentCount)
````

````python
### 12. Find maximum

# Write a Python program to find the larges in a list without using the built-in `max()` function.

list1 = [10,11,4,5,78]

maxValue = list1[0]
for i in list1:
    if i>maxValue:
        maxValue = i
    # print(i)
print("Maximum Value is:",maxValue)
````

````python
# Using While Loop 


list1 = [10,11,4,5,78,99]

i = 0
maxValue = list1[0]
while i<len(list1):
    # print(list1[i])
    if list1[i]>maxValue:
        maxValue = list1[i]
    i+=1
print("Maximum Value is:",maxValue)
````
