#### Important String Method

```python
#### Methods of strings


### .isalpha() : only alphabetical characters
### .isdigit() : only digit value check 0-9
### .isalnum() : it check alphabet or numerical value
### .isspace() : it check only space value
### .isupper() : upper case value
### .islower() : lower case value
```

```python
s = "hello world"

s1 = "123456789"

s2 = "abc123"

s3 = " "

s4 = "HELLO"


print(s.islower())
print(s1.isdigit())
print(s3.isspace())
print(s.isalnum())
print(s.islower())
print(s.isupper())
print(s4.isupper())
```
